
La aproximación es el núcleo del análisis numérico. Cuando
aproximamos la solución de un problema, reemplazamos un operador matemático complejo (como una derivada o una integral) por una aproximación discreta realizable por un ordenador.


## Fuentes de error 

Para evaluar la validez de cualquier aproximación numérica, debemos entender las fuentes del error. En términos generales, los errores se clasifican en:

- Errores de redondeo: Se originan debido a que las computadoras representan los números reales con una precisión finita (aritmética de punto flotante).

- Errores de truncamiento: Surgen al reemplazar un proceso matemático infinito (como una serie de Taylor o un límite) por un procedimiento finito.

- Errores inherentes: Son aquellos presentes en los datos de entrad del problema, causados por limitaciones en los instrumentos de medición física.


## Error absoluto, relativo y porcentual

Para medir cuantitativamente la desviación de nuestra aproximación respecto al valor real, definimos el error absoluto $(E_a)$ y el error relativo $(E_r)$.
### Error Absoluto\
Si definimos $x$ como el valor real y $\hat{x}$ o $x*$ como el valor aproximado, el error absoluto se expresa como:

$$ E_a = |x − \hat{x}| $$

Por ello, recurrimos al error relativo, el cual normaliza el error respecto al valor real:
### Error Relativo

$$E_r = \frac{|x − \hat{x}|}{|x|}$$



El error relativo se expresa frecuentemente como porcentaje, dando lugar al error relativo porcentual $(Er\%)$:

### Error relativo porcentual

$$ E_r\% = E_r \times 100\% = \frac{|x − \hat{x}|}{|x|} \times 100\% , x \neq 0 $$


Esta forma de expresar el error es particularmente útil en aplicaciones prácticas, ya que permite interpretar la precisión de la aproximación en términos de porcentaje.

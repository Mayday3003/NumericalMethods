
La aproximación es el núcleo del análisis numérico. Cuando
aproximamos la solución de un problema, reemplazamos un operador matemático complejo (como una derivada o una integral) por una aproximación discreta realizable por un ordenador.


## Fuentes de error 

Para evaluar la validez de cualquier aproximación numérica, debemos entender las fuentes del error. En términos generales, los errores se clasifican en:

- Errores de redondeo: Se originan debido a que las computadoras representan los números reales con una precisión finita (aritmética de punto flotante).

- Errores de truncamiento: Surgen al reemplazar un proceso matemático infinito (como una serie de Taylor o un límite) por un procedimiento finito.

- Errores inherentes: Son aquellos presentes en los datos de entrad del problema, causados por limitaciones en los instrumentos de medición física.


## Error absoluto, relativo y porcentual

Para medir cuantitativamente la desviación de nuestra aproximación respecto al valor real, definimos el error absoluto $(E_a)$ y el error relativo $(E_r)$.
### Error Absoluto
Si definimos $x$ como el valor real y $\hat{x}$ o $x*$ como el valor aproximado, el error absoluto se expresa como:

$$ E_a = |x − \hat{x}| $$

Por ello, recurrimos al error relativo, el cual normaliza el error respecto al valor real:
### Error Relativo

$$E_r = \frac{|x − \hat{x}|}{|x|}$$



El error relativo se expresa frecuentemente como porcentaje, dando lugar al error relativo porcentual $(Er\%)$:

### Error relativo porcentual

$$ E_r\% = E_r \times 100\% = \frac{|x − \hat{x}|}{|x|} \times 100\% , x \neq 0 $$


Esta forma de expresar el error es particularmente útil en aplicaciones prácticas, ya que permite interpretar la precisión de la aproximación en términos de porcentaje.





### Ejercicio 1
Sea $p = 0.34 \times 10^{-4}$ y $\hat{p} = 0.54 \times 10^{-4}$ . Determine el error absoluto y el error relativo.

```python
def ErrorAbsoluto(valor_real:float, valor_aproximado: float) -> float:
    error_absoluto = abs(valor_real - valor_aproximado)
    return error_absoluto



def ErrorRelativo(valor_real: float, valor_aproximado: float) -> float:
    error_relativo = abs(valor_real - valor_aproximado) / abs(valor_real)
    return error_relativo



def ErrorPorcentual(error_relativo) -> float:

    error_porcentual = error_relativo * 100
    return error_porcentual




def MostrarErrores(valor_real,valor_aproximado):
    error_absoluto = ErrorAbsoluto(valor_real, valor_aproximado)

    print(f'El Error absoluto es {error_absoluto}')
    print('----------------------------------------------------')

    error_relativo = ErrorRelativo(valor_real,valor_aproximado)
    print(f'El Error relativo es {error_relativo}')

    print('----------------------------------------------------')
    
    error_porcentual = ErrorPorcentual(error_relativo)
    print(f'El Error porcentual es {error_porcentual}')



valor_real = 0.34 * (10 ^ -4)
valor_aproximado = 0.54 * (10 ^ -4)

MostrarErrores(valor_real, valor_aproximado)
```

Para tener en cuenta
Por lo general, interesa el error absoluto y no el error relativo; pero,
cuando el valor exacto de una cantidad es “muy pequeno” o “muy
grande”, los errores relativos son mas significativos.

# Exactitud y precisión

Un algoritmo es numéricamente estable si los errores de redondeo
no se amplifican significativamente durante el proceso de calculo. Un
algoritmo inestable puede producir resultados sin sentido, incluso si el
problema matematico esta bien condicionado.
Los errores en los calculos y medidas se pueden caracterizar por la
exactitud y la precision.





## Exactitud
La exactitud se refiere a qué tan cercano esta el valor calculado o
medido del valor verdadero.

## Precision
La precision se refiere a qué tan cercanos se encuentran, unos de
otros, diversos valores calculados 0 medidos.


#### **Nota**
- Un algoritmo puede ser preciso pero inexacto (todos los resultados
estan cerca entre si, pero lejos del valor real). Esto suele deberse a
errores sistematicos 0 sesgos en el método.

- Un algoritmo puede ser exacto pero impreciso (el promedio de los
resultados es cercano al valor real, pero hay gran dispersién). Esto
indica errores aleatorios.

- El objetivo ideal es lograr un algoritmo exacto y preciso.

# Qué es la serie de taylor?
La serie de Taylor es una herramienta matematica importante
en el analisis y aproximacion de funciones. Esta serie se utiliza
para representar una funci6én mediante una suma infinita de
términos polinomiales, que estan centrados en un punto
especifico

## Para qué sirve?
La serie de Taylor sirve como una herramienta poderosa para
representar y aproximarse a funciones complicadas mediante
polinomios simples, lo que facilita el andlisis y la manipulacién
de estas funciones en una variedad de contextos matematicos
y cientificos.

# **Serie de Taylor**
Sea $f$ una función que tiene derivadas continuas hasta el orden $n+1$ en un intervalo $[a,b]$. Entonces, para cualquier $x,x_0 \in [a,b]$, existe un número $\xi$ entre $x$ y $x_0$ tal que:

$$
f(x)=P_n(x)+R_n(x)
$$

donde

$$
P_n(x)=\sum_{k=0}^{n}\frac{f^{(k)}(x_0)}{k!}(x-x_0)^k
$$

es el **polinomio de Taylor de grado $n$** alrededor de $x_0$, y

$$
R_n(x)=\frac{f^{(n+1)}(\xi)}{(n+1)!}(x-x_0)^{n+1}
$$

es el **residuo o error de truncamiento**.

## Polinomio de Taylor de grado 3

Queremos construir el polinomio de Taylor de grado $n=3$ de

$$
f(x)=\ln(x)
$$

alrededor de $x_0=1$.

El polinomio de Taylor es:

$$
P_3(x)=\sum_{k=0}^{3}\frac{f^{(k)}(x_0)}{k!}(x-x_0)^k
$$

Calculamos las derivadas:

$$
f(x)=\ln(x)
$$

$$
f'(x)=\frac{1}{x}
$$

$$
f''(x)=-\frac{1}{x^2}
$$

$$
f'''(x)=\frac{2}{x^3}
$$

Evaluamos en $x_0=1$:

$$
f(1)=0
$$

$$
f'(1)=1
$$

$$
f''(1)=-1
$$

$$
f'''(1)=2
$$

Por lo tanto:

$$
P_3(x)
=
0
+
\frac{1}{1!}(x-1)
-\frac{1}{2!}(x-1)^2
+\frac{2}{3!}(x-1)^3
$$

Simplificando:

$$
\boxed{
P_3(x)=(x-1)-\frac{(x-1)^2}{2}+\frac{(x-1)^3}{3}
}
$$

## Aproximación de $f(1.5)$

Utilizamos el polinomio:

$$
P_3(x)=(x-1)-\frac{(x-1)^2}{2}+\frac{(x-1)^3}{3}
$$

Para $x=1.5$:

$$
P_3(1.5)
=
(1.5-1)
-\frac{(1.5-1)^2}{2}
+\frac{(1.5-1)^3}{3}
$$

$$
P_3(1.5)
=
0.5-\frac{0.25}{2}+\frac{0.125}{3}
$$

$$
P_3(1.5)
=
0.5-0.125+0.0416667
$$

Por lo tanto:

$$
\boxed{P_3(1.5)\approx0.416667}
$$

El valor real es:

$$
\ln(1.5)\approx0.405465
$$

Por lo tanto, la aproximación mediante Taylor de grado 3 es:

$$
\boxed{\ln(1.5)\approx0.416667}
$$

## Aproximación de $f(10)$

Utilizamos:

$$
P_3(x)=(x-1)-\frac{(x-1)^2}{2}+\frac{(x-1)^3}{3}
$$

Para $x=10$:

$$
P_3(10)
=
(10-1)-\frac{(10-1)^2}{2}+\frac{(10-1)^3}{3}
$$

$$
P_3(10)
=
9-\frac{81}{2}+\frac{729}{3}
$$

$$
P_3(10)
=
9-40.5+243
$$

Por lo tanto:

$$
\boxed{P_3(10)=211.5}
$$

El valor real es:

$$
\ln(10)\approx2.302585
$$

La aproximación es extremadamente mala porque $x=10$ se encuentra muy lejos del punto de expansión $x_0=1$.

$$
\boxed{\ln(10)\approx211.5\quad\text{usando Taylor de grado 3}}
$$

## Polinomio de Taylor de grado 3 para $f(x)=e^x$

Queremos construir el polinomio de Taylor de grado $n=3$ alrededor de:

$$
x_0=1
$$

La función es:

$$
f(x)=e^x
$$

Sus derivadas son:

$$
f(x)=e^x
$$

$$
f'(x)=e^x
$$

$$
f''(x)=e^x
$$

$$
f'''(x)=e^x
$$

Evaluamos todas las derivadas en $x_0=1$:

$$
f(1)=e
$$

$$
f'(1)=e
$$

$$
f''(1)=e
$$

$$
f'''(1)=e
$$

El polinomio de Taylor de grado 3 es:

$$
P_3(x)
=
f(1)
+
f'(1)(x-1)
+
\frac{f''(1)}{2!}(x-1)^2
+
\frac{f'''(1)}{3!}(x-1)^3
$$

Sustituyendo los valores:

$$
P_3(x)
=
e+e(x-1)
+\frac{e(x-1)^2}{2!}
+\frac{e(x-1)^3}{3!}
$$

Por lo tanto:

$$
\boxed{
P_3(x)
=
e\left[
1+(x-1)+\frac{(x-1)^2}{2}
+\frac{(x-1)^3}{6}
\right]
}
$$

## Aproximación de $f(1.5)$

Usamos:

$$
P_3(x)
=
e\left[
1+(x-1)+\frac{(x-1)^2}{2}
+\frac{(x-1)^3}{6}
\right]
$$

Para $x=1.5$:

$$
P_3(1.5)
=
e\left[
1+(1.5-1)
+\frac{(1.5-1)^2}{2}
+\frac{(1.5-1)^3}{6}
\right]
$$

$$
P_3(1.5)
=
e\left[
1+0.5+\frac{0.25}{2}+\frac{0.125}{6}
\right]
$$

$$
P_3(1.5)
=
e(1+0.5+0.125+0.0208333)
$$

$$
P_3(1.5)
=
e(1.6458333)
$$

$$
\boxed{P_3(1.5)\approx4.473839}
$$

El valor real es:

$$
f(1.5)=e^{1.5}
$$

$$
\boxed{f(1.5)\approx4.481689}
$$

## Aproximación de $f(5)$

Para $x=5$:

$$
P_3(5)
=
e\left[
1+(5-1)
+\frac{(5-1)^2}{2}
+\frac{(5-1)^3}{6}
\right]
$$

$$
P_3(5)
=
e\left[
1+4+\frac{16}{2}+\frac{64}{6}
\right]
$$

$$
P_3(5)
=
e\left[
1+4+8+10.666667
\right]
$$

$$
P_3(5)
=
e(23.666667)
$$

$$
\boxed{P_3(5)\approx64.332670}
$$

El valor real es:

$$
f(5)=e^5
$$

$$
\boxed{f(5)\approx148.413159}
$$

## Resultados

El polinomio obtenido es:

$$
\boxed{
P_3(x)
=
e\left[
1+(x-1)+\frac{(x-1)^2}{2}
+\frac{(x-1)^3}{6}
\right]
}
$$

Comparando el valor real con la aproximación:

| $x$ | Dato real $f(x)=e^x$ | Dato aproximado $P_3(x)$ |
|---:|---:|---:|
| $1.5$ | $4.481689$ | $4.473839$ |
| $5$ | $148.413159$ | $64.332670$ |
| $10$ | $22026.465795$ | $467.544474$ |

Se observa que para $x=1.5$, que está relativamente cerca del punto de expansión $x_0=1$, la aproximación es bastante buena.

En cambio, para $x=5$ y especialmente para $x=10$, el error aumenta considerablemente porque estos valores están cada vez más alejados de $x_0=1$.

Esto demuestra que un polinomio de Taylor de grado bajo aproxima mejor la función cerca del punto donde se construyó.
